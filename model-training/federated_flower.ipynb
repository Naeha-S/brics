## 2. FedAvg Aggregation with Differential Privacy $(\epsilon=2.1)$
Defines client update weighting $W_{t+1} = W_t + \sum \frac{n_k}{N} \Delta W_k$ and gradient noise addition.

```python
total_samples = sum(n['samples'] for n in nodes.values())
dp_params = {'epsilon': 2.1, 'delta': 1e-5, 'clip_norm': 1.0}

print('Federation Weights per Node:')
for code, data in nodes.items():
    weight = data['samples'] / total_samples
    print(f'  [{code}] {data["jurisdiction"]:15s} : {weight:.4f} ({data["samples"]:,} samples)')
print(f'Total Federation Pool: {total_samples:,} samples under Differential Privacy ε={dp_params["epsilon"]}')
```
